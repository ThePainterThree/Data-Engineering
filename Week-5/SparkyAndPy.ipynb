```python
from pyspark.sql import SparkSession
import string

spark = SparkSession.builder \
    .appName("RDD_Practice") \
    .getOrCreate()

sc = spark.sparkContext
    
text = sc.textFile("/Users/Bootcamp/Desktop/Labs/sample.txt")
stop_list = sc.textFile("/Users/Bootcamp/Desktop/Labs/stop_words.txt")
stop_words = set(stop_list.map(lambda x: x.strip().lower()).collect())

text_to_lowercase = text.flatMap(lambda x: x.lower().split())

clean_words = text_to_lowercase.map(lambda x: x.strip(string.punctuation))

filtered_words = clean_words.filter(lambda x: x not in stop_words)

word_count = filtered_words.map(lambda x: (x, 1)).reduceByKey(lambda a, b: a + b)

top_5 = word_count.takeOrdered(5, key=lambda x: -x[1])

top_5
```
